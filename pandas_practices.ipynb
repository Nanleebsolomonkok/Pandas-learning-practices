```python
from typing_extensions import AsyncGenerator
from google.colab import files
upload = files.upload()
import pandas as pd
#DaataFram is a table
#series is a list
try0 = pd.DataFrame({ 'yes': [131, 67], 'No': [35,68]})
try1 = pd.DataFrame({'Dob': ['I am 21 year old', 'university of mines', 'NLP'],
              'Level':['I am in level 200', 'Computer science', 'USA FOR PHD'],
              'courses':['java', 'python', 'SQL']},
              index =[1, 2,3])

pd.Series([1,2,3,4,'5'], index=['A', 'B', 'C', 'D', 'D'], name = 'Alphabets' )
#csv fully means --- commas seperated values
# print(try0.head())
# print(try1.head())


#READING CSV
wine_reviews = pd.read_csv("titanic.csv")
print(wine_reviews.head())
# print(wine_reviews.shape)
# print(wine_reviews.Calories)
# print(wine_reviews['Calories'])
# print(wine_reviews['Calories'].head())

#indexing
#iLoc means index based section

# : means to select everything
wine_reviews.iloc[:,0]
wine_reviews.iloc[:3, 0]
# you can also pass a list
wine_reviews.iloc[[1,2,3,4], 0]
 #when you used a negative index you it start from the last index
wine_reviews.iloc[-5:]
#loc

#Loc means lable based section
wine_reviews.loc[0]
print(wine_reviews.loc[:, ['Name', 'Age']])
wine_reviews.set_index('Name')
print(wine_reviews.Sex =='male')
print(wine_reviews.loc[wine_reviews.Sex =='male'])
#conditions section
# supoosing we want to determine two different conditions. we used the ampersand(&)
wine_reviews.loc[(wine_reviews.Sex =='male') & (wine_reviews.Embarked == 'S')]

#supose you want to determine two condition for which any of them is meant we use pipe(|)
wine_reviews.loc[(wine_reviews.Sex =='male') | (wine_reviews.Embarked == 'S')]
#to select a particular record which is in your table we osed (isin)
wine_reviews.loc[wine_reviews.Embarked.isin(['S'])]
#To be able to determine empty records we use isnull() or notnull()
print(wine_reviews.loc[wine_reviews.Embarked.isnull()])
wine_reviews.loc[wine_reviews.Embarked.notnull()]
#Assigning data
# wine_reviews['Critical'] = 'everyone'
# wine_reviews['Critical']

#summary fuction amd Maps
print(wine_reviews)
wine_reviews.Pclass.describe()
print(wine_reviews.Pclass.mean())
wine_reviews.Pclass.unique()
wine_reviews.Pclass.value_counts()

#map
wine_reviews_pclass_mean = wine_reviews.Pclass.mean()
wine_reviews.Pclass.map(lambda p : p - wine_reviews_pclass_mean)

#apply
def remean_Pclass(row):
  row.Pclass = row.Pclass - wine_reviews_pclass_mean
  return row
wine_reviews.apply(remean_Pclass, axis = 'columns')

wine_reviews.Name + " - "+ wine_reviews.Sex
#Grouping
wine_reviews.groupby('Survived').Survived.count()
wine_reviews.Survived.value_counts()
wine_reviews.PassengerId.value_counts()
wine_reviews.groupby('Fare').Fare.max()

wine_reviews.groupby('Sex').apply(lambda df : df.Name.iloc[0])
wine_reviews.groupby(['Sex', 'Survived']).apply(lambda df : df.loc[df.Fare.idxmax()])

wine_reviews.groupby(['Sex']).Fare.agg([len, min, max])
#multi_index
multi = wine_reviews.groupby(['Sex']).Fare.agg([len])
multi

#coverting in to regular index
#sorting
sorting = multi.reset_index()
sorting.sort_values(by ='len', ascending=False)
sorting.sort_index(ascending=False)
sorting.sort_values(by =['Sex', 'len'])

#DataTypes
wine_reviews.dtypes
#coverting one datatype to another
wine_reviews.Fare.astype('int64')

#missing values
wine_reviews[pd.isnull(wine_reviews.Fare)]
wine_reviews[pd.notnull(wine_reviews.Fare)]
#To change NaN to different name
wine_reviews.Age.fillna('unknown')
wine_reviews.Embarked.replace('@NaN', '@unknown')
wine_reviews.Name.replace('John Doe', 'Nanleeb Solomon')
#remove missing values
wine_reviews.dropna()
wine_reviews.isnull().sum()
pd.isnull(wine_reviews.Embarked)

wine_reviews
#Renaming
wine_reviews.rename(columns = {'Sex': 'Gender'})
wine_reviews.rename(index={0: 'kok', 1: 'nanleeb'})
wine_reviews.rename_axis('wines', axis='rows').rename_axis('field', axis='columns')
#data = pd.read_csv('titanic.csv')
#combing
data = pd.read_csv('data.csv')
#combing
#concat()
pd.concat([wine_reviews, data])
#join
wine_reviews.join(data)
wine_reviews.merge(data, left_on ='Pclass', right_on = 'Maxpulse')


```
